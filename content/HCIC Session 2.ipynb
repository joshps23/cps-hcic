## Slide 4 — `elif` — More Than Two Choices 🎚️

### How `elif` Works

`elif` means **"else if"**. Use it when you need to check **more than two** possibilities.

Python checks each condition **top to bottom** and runs the **FIRST one that's true**.
`else` catches everything that's left.

### Grading Example

> ⚠️ **Order matters** — put the strictest test first!

````python
score = 82

if score >= 90:
    print("Grade: A")
elif score >= 75:
    print("Grade: B")
elif score >= 50:
    print("Grade: C")
else:
    print("Keep practicing!")

# Output: Grade: B
````

## Slide 6 — Building Smarter Conditions 🧠

### Comparison Operators

Compare two values. Each one gives back **`True`** or **`False`**.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 3` | `True` |
| `>` | greater than | `5 > 3` | `True` |
| `<` | less than | `5 < 3` | `False` |
| `>=` | greater than or equal | `5 >= 5` | `True` |
| `<=` | less than or equal | `5 <= 4` | `False` |

Run the cell to see them in action:

````python
print(5 == 5)
print(5 != 3)
print(5 > 3)
print(5 < 3)
print(5 >= 5)
print(5 <= 4)
````

### Logic Operators

Combine conditions with **`and`**, **`or`**, **`not`**.

- `and` → both conditions must be True
- `or` → at least one must be True
- `not` → flips True to False (and vice versa)

````python
age = 12
has_ticket = True

print(age >= 10 and has_ticket)   # True
print(age < 5 or has_ticket)      # True
print(not has_ticket)             # False
````

## Slide 8 — Shortcut Operators: `+=`, `-=`, `*=`, `/=` ⚡

### Update a Variable Using Itself

Programmers often update a variable based on its **current value**.

Compound operators are a shortcut: `x += 5` does the same thing as `x = x + 5`, just shorter and easier to read!

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+=` | Add and assign | `score = 10; score += 5` | `score` is `15` |
| `-=` | Subtract and assign | `lives = 3; lives -= 1` | `lives` is `2` |
| `*=` | Multiply and assign | `coins = 4; coins *= 3` | `coins` is `12` |
| `/=` | Divide and assign | `energy = 100; energy /= 4` | `energy` is `25.0` |

Run each one to see them in action:

````python
score = 10
score += 5
print("score:", score)

lives = 3
lives -= 1
print("lives:", lives)

coins = 4
coins *= 3
print("coins:", coins)

energy = 100
energy /= 4
print("energy:", energy)
````

### 🔁 Real Use: Counting in a Loop

````python
total = 0
for n in [10, 20, 30]:
    total += n        # total grows: 10, 30, 60
    print("total is now:", total)
````

## Slide 12 — Loops: Beyond the Basics 🚀

### `range(start, stop, step)`

Control where loops start, stop, and how big each step is.

````python
for i in range(2, 10, 2):
    print(i)

# 2, 4, 6, 8
````

### Nested Loops — a loop *inside* a loop

````python
for row in range(3):
    for col in range(3):
        print(row, col)
````

### `break` & `continue`

- **`break`** → stops the loop **early**
- **`continue`** → skips to the **next** round

````python
for n in range(5):
    if n == 3:
        break
    print(n)

# 0, 1, 2
````

````python
# continue example - skip the number 3
for n in range(5):
    if n == 3:
        continue
    print(n)

# 0, 1, 2, 4
````

## Slide 13 — ⚠️ Common Loop Pitfalls

### Off-by-One Trap

`range(1, 5)` gives `1, 2, 3, 4` — **NOT** `5`. The stop value is **excluded**!

````python
for i in range(1, 5):
    print(i)

# Prints: 1 2 3 4
# Missing 5!
````

### ⚠️ Changing a List While Looping

Modifying a list **during** iteration can skip items or cause errors.

````python
nums = [1, 2, 3, 4]
for n in nums:
    if n == 2:
        nums.remove(n)
print(nums)
# Watch out — number 3 might be skipped!
````

## Slide 14 — List Slicing ✂️: Picking Parts

Slicing lets you grab a **chunk** of a list (or string) using `list[start:stop]`.

| Slice | What it does |
|---|---|
| `[1:4]` | Items from index `1` up to (but **not including**) index `4` |
| `[:3]` | First three items (start to index 3) |
| `[2:]` | From index `2` to the end |

````python
friends = ["Anna", "Ben", "Charlie", "Dana", "Eli"]
print(friends[1:4])  # ['Ben', 'Charlie', 'Dana']
print(friends[:3])   # ['Anna', 'Ben', 'Charlie']
print(friends[2:])   # ['Charlie', 'Dana', 'Eli']
````

## Slide 15 — Slicing: More Powerful Tricks 🪄

### Negative Indices

Count from the **end**. `-1` is the last item, `-2` is second to last.

````python
nums = [10, 20, 30, 40, 50]
print(nums[-1])   # 50
print(nums[-3:])  # [30, 40, 50]
````

### Step Value

`list[start:stop:step]` — **jump** by step. Skip every other item!

````python
nums = [1, 2, 3, 4, 5, 6]
print(nums[::2])    # [1, 3, 5]  (every 2nd, starting from 0)
print(nums[1::2])   # [2, 4, 6]  (every 2nd, starting from 1)
````

### 🔄 Reverse a List

Use a step of `-1` to flip the whole list backwards.

````python
word = "python"
print(word[::-1])   # 'nohtyp'

nums = [1, 2, 3]
print(nums[::-1])   # [3, 2, 1]
````

## Slide 16 — Slicing Pitfalls & Practice ⚠️

### Slice vs Index

`list[2]` returns **ONE item**. `list[2:3]` returns a **LIST** with one item. Different types!

````python
nums = [10, 20, 30, 40]
print(nums[2])      # 30 (a number)
print(nums[2:3])    # [30] (a list)
````

### Out-of-Range Is OK (for slicing)

**Slicing** past the end won't crash — it just stops.
But **indexing** past the end DOES crash!

````python
nums = [1, 2, 3]
print(nums[0:99])   # [1, 2, 3]  fine!
# print(nums[99])   # ❌ IndexError! (uncomment to see the error)
````

## Slide 17 — While Loops 🔄: Repeat Until Done

A `while` loop keeps going **as long as a condition is True**.

### How it works:
1. **Ask** a question (get input)
2. **Check** if the answer is correct
3. **Repeat** if wrong — until they get it right!

> ⚠️ **Watch out:** if your condition is **never** False, the loop runs **forever**!
> If that happens, click the **▢ Interrupt** button (Stop) at the top of Jupyter.

The example below keeps asking until you type `python`:

````python
guess = ""
while guess.lower() != "python":
    guess = input("Guess the magic word: ")
print("You guessed it! 🎉")
````
