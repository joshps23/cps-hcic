## Slide 5 — Getting User Input 🎤

### Ask a Question

The `input()` function lets you **ask questions and get answers** from the person using your program!

When you run the cell, a little box will pop up — type your answer and press Enter.

````python
name = input("What is your name? ")
print("Nice to meet you, " + name + "!")
````

### 💡 Notice the `+` symbol?

We use `+` to **join** (or "stick together") pieces of text.

So `"Nice to meet you, " + name + "!"` becomes `"Nice to meet you, Joshua!"` if you typed Joshua.
