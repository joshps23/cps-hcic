## Slide 3 — Quick Recap: While Loops 🔄

## Slide 4 — Quick Recap: Themed Quiz Game 🏆

Combine **lists** + **for loops** + **if/else** to make a quiz.

````python
questions = ["What is the Red Planet?", "Sun is a star? (yes/no)"]
answers   = ["mars", "yes"]
score = 0

for i in range(len(questions)):
    response = input(questions[i] + " ")
    if response.lower() == answers[i]:
        print("Correct! ✅")
        score += 1
    else:
        print("Wrong! ❌")

print("Final Score:", score)
````

## Slide 7 — List Slicing ✂️: Picking Parts

Slicing grabs a **chunk** of a list using `list[start:stop]`.

| Slice | What it does |
|---|---|
| `[1:4]` | Items from index `1` up to (but **not including**) `4` |
| `[:3]` | First three items |
| `[2:]` | From index `2` to the end |

````python
friends = ["Anna", "Ben", "Charlie", "Dana", "Eli"]
print(friends[1:4])   # ['Ben', 'Charlie', 'Dana']
print(friends[:3])    # ['Anna', 'Ben', 'Charlie']
print(friends[2:])    # ['Charlie', 'Dana', 'Eli']
````

## Slide 8 — Slicing: More Powerful Tricks 🪄

### Negative Indices
Count from the end. `-1` = last item, `-2` = second to last.

````python
nums = [10, 20, 30, 40, 50]
print(nums[-1])   # 50
print(nums[-3:])  # [30, 40, 50]
````

### Step Value
`list[start:stop:step]` — jump by step.

````python
nums = [1, 2, 3, 4, 5, 6]
print(nums[::2])    # [1, 3, 5]  every 2nd from index 0
print(nums[1::2])   # [2, 4, 6]  every 2nd from index 1
````

### 🔄 Reverse a List with Step `-1`

````python
word = "python"
print(word[::-1])   # 'nohtyp'

nums = [1, 2, 3]
print(nums[::-1])   # [3, 2, 1]
````

## Slide 9 — Slicing 2D & 3D Lists 🧊

### 2D Lists (Grids)
A list of lists! Use **two** index numbers: `grid[row][col]`.

Imagine this 3×3 grid:

| | col 0 | col 1 | col 2 |
|---|---|---|---|
| **row 0** | 1 | 2 | 3 |
| **row 1** | 4 | 5 | **6** ← `grid[1][2]` |
| **row 2** | 7 | 8 | 9 |

````python
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(grid[1][2])   # 6 — row 1, column 2
````

### Rows & Columns

````python
# Grab rows 0 and 1 (the top two)
print(grid[0:2])
# [[1, 2, 3], [4, 5, 6]]

# Grab column 0 with a loop
col = [row[0] for row in grid]
print(col)
# [1, 4, 7]
````

### 3D Lists (Cubes)
**Two stacked grids**. Use THREE index numbers: `cube[layer][row][col]`.

````python
cube = [
    [[1, 2], [3, 4]],   # layer 0
    [[5, 6], [7, 8]]    # layer 1
]
print(cube[1][0][1])   # 6 — layer 1, row 0, col 1
````

## Slide 10 — Slicing Pitfalls & Practice ⚠️

### Slice vs Index

`list[2]` returns **ONE item**. `list[2:3]` returns a **LIST** with one item. Different types!

````python
nums = [10, 20, 30, 40]
print(nums[2])      # 30 (a number)
print(nums[2:3])    # [30] (a list)
````

### Out-of-Range Is OK for Slicing
**Slicing** past the end won't crash — it just stops.
But **indexing** past the end DOES crash!

````python
nums = [1, 2, 3]
print(nums[0:99])   # [1, 2, 3]  fine!
# print(nums[99])   # ❌ IndexError! (uncomment to see)
````

## Slide 12 — Functions 🧰: Reusable Commands

### What Are Functions?
Functions are like **mini-programs** you can reuse anywhere.

They save time and make code easier to read.

**Anatomy:**
- `def` starts the function
- `greet` is the name
- `(name)` is the **parameter** (input)
- The indented block is what it DOES

## Slide 13 — Functions: Level Up 🆙

### Multiple Parameters
Functions can take **more than one** input.

````python
def greet(name, mood):
    print("Hi " + name + "!")
    print("Hope you feel " + mood + ".")

greet("Ash", "great")
greet("Maya", "calm")
````

### Return a Value
Use `return` to send a result **back** so you can use it later.

````python
def add(a, b):
    return a + b

total = add(5, 3)
print("The total is", total)

# You can use the result directly too
print(add(10, 20))
````

### 💡 `print` vs `return`

## Slide 15 — Try and Except 🛡️: Handling Mistakes

### What is Try/Except?

Try/except is a **safety net** — it catches errors before they crash your program!

**Without it:** typing letters when the program wants numbers → CRASH 💥
**With it:** you can print a friendly message instead 😊

````python
try:
    number = int(input("Enter a number: "))
    print(number * 2)
except:
    print("Oops! That's not a number!")
````

### How it works
1. Python **TRIES** to run the code in the `try:` block
2. If something goes wrong, Python **JUMPS** to the `except:` block
3. If nothing goes wrong, `except:` is skipped

## Slide 16 — Try / Except: Level Up 🆙

You can catch **specific** error types with different messages.

### Catch Specific Errors

````python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)
except ValueError:
    print("That wasn't a number!")
except ZeroDivisionError:
    print("You can't divide by zero!")
````

### Add `else` and `finally`

- **`else`** → runs only if **NOTHING broke**
- **`finally`** → **ALWAYS** runs, no matter what

````python
try:
    age = int(input("Your age? "))
except ValueError:
    print("Numbers only, please!")
else:
    print("Next year you'll be", age + 1)
finally:
    print("Thanks for trying!")
````

## Slide 18 — Dictionaries 📖: Key → Value

### What Are Dictionaries?

Dictionaries store information as **key-value pairs**.

Just like a real dictionary:
- The **word** is the **key**
- The **meaning** is the **value**

**Syntax:** uses **curly braces `{ }`** with `key: value` pairs.

````python
pokemon = {"Pikachu": "Electric", "Charmander": "Fire"}
print(pokemon["Pikachu"])     # Electric
print(pokemon["Charmander"])  # Fire
````

## Slide 19 — Dictionaries: Level Up 🆙

### Add, Update, and Check Keys
Dictionaries can **grow and change** as your program runs.

````python
scores = {"Ash": 85, "Maya": 92}

scores["Sam"] = 78     # ADD a new key
scores["Ash"] = 90     # UPDATE an existing key

if "Maya" in scores:
    print("Maya scored", scores["Maya"])

print(scores)
````

### Loop Through a Dictionary
Use `.items()` to walk through every pair.

````python
sounds = {
    "cat": "meow",
    "dog": "woof",
    "cow": "moo"
}

for animal, noise in sounds.items():
    print(animal, "says", noise)
````
